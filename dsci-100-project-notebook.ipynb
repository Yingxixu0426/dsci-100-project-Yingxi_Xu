```python
import altair as alt
```

```python
plot1 = alt.Chart(data_filtered).mark_bar().encode(
    x=alt.X("reg_wins_pct:Q", bin=True, title="Regional Win Percentage"),
    y=alt.Y("count():Q", title="Number of Teams")
).properties(
    title="Distribution of Regional Win Percentage"
)

plot1
```

```python
finish_prop = data_filtered["tourney_finish"].value_counts(normalize=True).reset_index()
finish_prop.columns = ["tourney_finish", "proportion"]
plot2 = alt.Chart(finish_prop).mark_bar().encode(
    x=alt.X("tourney_finish:N", title="Tournament Finish"),
    y=alt.Y("proportion:Q", title="Proportion")
).properties(
    title="Proportion of Teams by Tournament Finish"
)
plot2.configure_axisX(labelAngle=-45)
```

Visualization 1: %%% EDIT THIS MARKDOWN CELL%%%

```python
mean
finish_order = [
    "opening_round_loss",
    "first_round_loss",
    "second_round_loss",
    "top_16_loss",
    "top_8_loss",
    "top_4_loss",
    "top_2_loss",
    "champ"
]

mean["tourney_finish"] = pd.Categorical(
    mean["tourney_finish"],
    categories=finish_order,
    ordered=True
)

mean = mean.sort_values("tourney_finish")
mean
```

```python
mean_order_plot=alt.Chart(mean).mark_bar().encode(
    x=alt.X("tourney_finish").title("The final grade in the tourney"),
    y=alt.Y("mean_reg_wins_pct").title("Mean win percent in the region match(%)")
).configure_axisX(labelAngle=-45)
mean_order_plot
```
